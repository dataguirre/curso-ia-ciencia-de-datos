## Actividad 1: Modelar una montaña rusa

### Objetivo

Escribir una función `modelar_montana_rusa(historial)` que reciba el historial de movimientos de una montaña rusa —una lista de strings con los valores `"sube"` o `"baja"`— y devuelva la probabilidad estimada de cada movimiento.

### Requisitos

1. La función recibe una lista de strings. Cada elemento es `"sube"` o `"baja"`.
2. La probabilidad de cada movimiento se estima con su **frecuencia relativa** en el historial:

   $$P(\text{movimiento}) = \frac{\text{veces que aparece}}{\text{total de movimientos}}$$

3. Devuelve un diccionario con la forma `{"sube": p1, "baja": p2}`, donde `p1 + p2 = 1`.
4. Si el historial está vacío, no hay información previa: se asume que ambos movimientos son igual de probables (0.5 y 0.5).
5. Si aparece un valor distinto de `"sube"` o `"baja"`, se lanza un error.

### Ejemplos

| Historial | Resultado esperado |
|---|---|
| `["sube", "sube", "baja", "baja"]` | `{"sube": 0.5, "baja": 0.5}` |
| `["sube", "sube", "sube", "baja"]` | `{"sube": 0.75, "baja": 0.25}` |
| `["baja"]` | `{"sube": 0.0, "baja": 1.0}` |
| `[]` | `{"sube": 0.5, "baja": 0.5}` |


### Tareas
#### Tarea 1: Completar la función

````python
def modelar_montana_rusa(historial: list) -> dict:
  """Estima P(sube) y P(baja) por frecuencia relativa en el historial."""

  # TODO: completar función que cumpla con los requisitos
  sube, baja = 0, 0
  for movimiento in historial:
    if movimiento == "sube":
      sube += 1
    elif movimiento == "baja":
      baja += 1
    else:
      raise ValueError("Movimiento inválido")
  total = sube + baja
  if total == 0:
    return {"sube": 0.5, "baja": 0.5}
  return {"sube": sube/total, "baja": baja/total}
````

#### Tarea 2: Preparar texto

Implementar la función `tokenizar` que se encarga de los pasos 1 y 2: convierte una oración en una lista de tokens, en minúsculas y con los marcadores de inicio (`<s>`) y fin (`</s>`).

Ejemplos:

```
tokenizar("Quiero ir al cine") == ["quiero", "ir", "al", "cine"]
tokenizar("Hola! soy Daniel") == ["hola", "soy", "daniel"]

```



````python
import re

PATRON = r"[a-záéíóúüñ]+"   # palabras en minúscula, sin puntuación ni números
INICIO, FIN = "<s>", "</s>"


def tokenizar(oracion: str) -> list:
    """Parte la oración en palabras minúsculas, entre los marcadores <s> y </s>."""

    # TODO: pasar la oración a minúsculas y extraer las palabras con PATRON
    #       Pista: re.findall(PATRON, texto) devuelve la lista de palabras.
    clean_text = re.findall(PATRON, oracion.lower())
    # TODO: devolver la lista con INICIO al principio y FIN al final
    return [INICIO] + clean_text + [FIN]
````

#### Tarea 3: Entrenar el modelo

Pasos 3 y 4: cuenta cuántas veces cada palabra sigue a cada contexto. Devuelve un diccionario de contadores:
```
bigramas["quiero"]  ->  Counter({"ir": 2, "comer": 1})
```

Con el corpus `["quiero ir al cine", "quiero ir al parque", "quiero comer algo"]`:

```
bigramas["quiero"]["ir"] == 2
bigramas["<s>"]["quiero"] == 3
```

````python
from collections import defaultdict, Counter

def entrenar_bigrama(corpus: list) -> dict:
    """Cuenta cuántas veces cada palabra sigue a cada contexto: {contexto: Counter}."""

    bigramas = defaultdict(Counter)

    # TODO: para cada oración del corpus, tokenizarla
    for oracion in corpus:
        tokens = tokenizar(oracion)
        # TODO: recorrer sus parejas (anterior, siguiente) y sumar 1 al contador
        for anterior, siguiente in zip(tokens, tokens[1:]):
            bigramas[anterior][siguiente] += 1

    return bigramas
````

#### Tarea 4: Calcular probabilidades

Paso 5: convierte los conteos de un contexto en probabilidades.

`probabilidades(bigramas, "quiero")  ->  {"ir": 0.67, "comer": 0.33}`

Reglas:
  - Las probabilidades deben sumar 1.
  - Si el contexto nunca se vio en el corpus, devuelve un diccionario vacío `{}`.

````python
def probabilidades(bigramas: dict, contexto: str) -> dict:
    """Convierte los conteos de un contexto en probabilidades que suman 1."""

    # TODO: obtener el Counter del contexto (ojo: puede no existir)
    dicc = bigramas.get(contexto, {})
    # TODO: dividir cada conteo entre el total y devolver el diccionario
    total = sum(dicc.values())
    return {key: value/total for key, value in dicc.items()}
````

#### Tarea 5: Predecir
Predice las `k` palabras más probables después de un texto.

El contexto de un bigrama es SOLO la última palabra del texto:

`predecir(bigramas, "quiero ir al")`  ->  usa el contexto "al"

Devuelve una lista de tuplas (palabra, probabilidad) ordenada de mayora menor probabilidad:

`[("cine", 0.5), ("parque", 0.5)]`

Casos especiales:
  - Si el texto está vacío, el contexto es INICIO (`<s>`).
  - Si el contexto no se vio en el corpus, devuelve una lista vacía `[]`.

````python
def predecir(bigramas: dict, texto: str, k: int = 3) -> list:
    """Devuelve las k palabras más probables tras el texto, como (palabra, probabilidad)."""

    # TODO: extraer las palabras del texto y tomar la última como contexto
    contexto = texto.split(" ")[-1]
    if not contexto:
        contexto = "<s>"
    # TODO: pedir sus probabilidades y devolver las k mayores, ya ordenadas
    dicc = probabilidades(bigramas, contexto)
    if not dicc:
        return []
    return sorted(dicc.items(), key=lambda x: x[1], reverse=True)[:k]
````

````python
# @title
# Celda de validación. No modificar.
from math import isclose


def _validar_predecir():
    for nombre in ("tokenizar", "entrenar_bigrama", "probabilidades", "predecir"):
        if nombre not in globals():
            print(f"✗ Todavía no existe '{nombre}'. Ejecuta las celdas anteriores.")
            return

    corpus_prueba = [
        "quiero ir al cine",
        "quiero ir al parque",
        "quiero comer algo",
    ]
    print(f"Corpus de prueba: {corpus_prueba}\n")
    b = entrenar_bigrama(corpus_prueba)
    # contexto "a": x tres veces, y dos veces, z una vez  -> sirve para probar el orden
    orden = entrenar_bigrama(["a x", "a x", "a x", "a y", "a y", "a z"])

    fallas = []

    def igual(a, b_):
        if isinstance(a, (list, tuple)) and isinstance(b_, (list, tuple)):
            return len(a) == len(b_) and all(igual(x, y) for x, y in zip(a, b_))
        if isinstance(a, (int, float)) and isinstance(b_, (int, float)):
            return isclose(a, b_)
        return a == b_

    def revisar(descripcion, calcular, esperado):
        try:
            obtenido = calcular()
        except Exception as e:
            obtenido = f"{type(e).__name__}: {e}"
        if igual(obtenido, esperado):
            print(f"✓ {descripcion}")
        else:
            print(f"✗ {descripcion}\n    esperaba: {esperado}\n    obtuvo:   {obtenido}")
            fallas.append(descripcion)

    revisar("usa solo la última palabra como contexto ('quiero ir al' -> 'al')",
            lambda: sorted(predecir(b, "quiero ir al", k=2)),
            [("cine", 0.5), ("parque", 0.5)])

    revisar("texto de una sola palabra",
            lambda: predecir(b, "quiero", k=2),
            [("ir", 2 / 3), ("comer", 1 / 3)])

    revisar("texto vacío usa el contexto <s>",
            lambda: predecir(b, "", k=1),
            [("quiero", 1.0)])

    revisar("contexto nunca visto -> lista vacía",
            lambda: predecir(b, "jirafa"), [])

    revisar("ordena de mayor a menor probabilidad",
            lambda: predecir(orden, "a"),
            [("x", 0.5), ("y", 1 / 3), ("z", 1 / 6)])

    revisar("k recorta la lista",
            lambda: predecir(orden, "a", k=1),
            [("x", 0.5)])

    revisar("k mayor que las opciones devuelve todas",
            lambda: len(predecir(orden, "a", k=99)), 3)

    revisar("devuelve tuplas (palabra, probabilidad)",
            lambda: all(isinstance(par, tuple) and len(par) == 2
                        and isinstance(par[0], str) and isinstance(par[1], float)
                        for par in predecir(orden, "a")),
            True)

    print()
    if fallas:
        n = len(fallas)
        print(f"Falta{'n' if n > 1 else ''} {n} prueba{'s' if n > 1 else ''}. "
              "Revisa los casos con ✗ antes de seguir.")
    else:
        print("Todas las pruebas de 'predecir' pasaron.")


_validar_predecir()
````

### Preguntas para discutir

**1. ¿Por qué hay contextos donde una palabra tiene probabilidad 1.00? ¿Confiarías en esa predicción?**

**R.** Pasa cuando el contexto solo tuvo una continuación en el corpus.

No hay que confiar: el modelo imprime igual un 1.00 respaldado por 500 casos y uno respaldado por 1. Casi siempre es contexto raro, no una regla del idioma. Además implica darle probabilidad 0 a todo lo demás.

Solución: mirar el conteo del contexto y aplicar
suavizado.

**2. ¿Qué pasa con `predecir` si escribes una palabra que no está en el corpus? ¿Cómo lo arreglaría un teclado real?**

**R.** `probabilidades` devuelve `{}` y `predecir` devuelve `[]`:
el teclado se queda mudo. Ojo con la causa más común aquí, que no es "out of vocabulary" real: `predecir`
usa `texto.split(" ")[-1]` sin normalizar, así que "Los NIÑOS SON" falla aunque el modelo
sí conozca "son". Un teclado real usa *backoff* al unigrama, un token `<UNK>` y un modelo personal que aprende tus palabras.

**3. Si en vez de bigramas usaras trigramas, ¿las predicciones serían mejores? ¿Qué se pierde?**

**R.**  Más contexto = predicciones más precisas *si hay datos*. Pero los
contextos posibles crecen como $V^{n-1}$ y el corpus no crece: casi todos los trigramas se
ven 0 o 1 vez, así que aumentan los huecos y los 1.00 sin evidencia, y el modelo memoriza
oraciones en vez de generalizar. Con Tatoeba (oraciones de ~7 palabras) el trigrama puro
probablemente empeore. Lo que se hace en la práctica es combinarlos: interpolación o
backoff entre trigrama, bigrama y unigrama.
